```python
import numpy as np

s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8, 7, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])

valid = True

for i in range(9):
    if set(s[i, :]) != set(range(1, 10)):
        valid = False
    if set(s[:, i]) != set(range(1, 10)):
        valid = False

for i in range(0, 9, 3):
    for j in range(0, 9, 3):
        box = s[i:i+3, j:j+3]
        if set(box.flatten()) != set(range(1, 10)):
            valid = False

print("Valid Sudoku" if valid else "Invalid Sudoku")
```

```python
import numpy as np

# Columns: [Age, Math Marks, Science Marks]
data = np.array([
    [18, 85, 78],
    [19, 92, 88],
    [17, 76, 95],
    [18, 65, 70],
    [20, 90, 85]
])
```

```python
#Get the shape of the matrix.
print(data.shape)
```

```python
#Find the average age of students.
print(np.mean(data[:, 0]))
```

```python
#Extract Math marks of all students.
print(data[:, 1])
```

```python
#Find the highest Science mark.
print(np.max(data[:, 2]))
```

```python
#Get details of the student who scored more than 90 in Math.
print(data[data[:, 1] > 90])
```

```python
#Increase Math marks of all students by 5.
data[:, 1] += 5
print(data)
```

```python
#Find how many students are younger than 19.
print(np.sum(data[:, 0] < 19))
```

```python
#Calculate the average marks in each subject (column-wise mean).
print(np.mean(data[:, 1:], axis=0))
```

```python
#Get data of students who scored at least 80 in both subjects.
print(data[(data[:, 1] >= 80) & (data[:, 2] >= 80)])
```

```python
#Replace all Science marks < 75 with 0
data[data[:, 2] < 75, 2] = 0
print(data)
```
